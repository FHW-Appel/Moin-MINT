```python
variable1 = -6
while variable1 != 0:
    if variable1 > 0:
        variable1 = variable1 - 1
    else:
        variable1 = variable1 + 1
    print(variable1)
```

### Funktionen

```python
def vergleiche(a, b):
    if a < b:
        print(a, "ist kleiner als", b)
    else:
        print(a, "ist glößer oder gleichgroß wie", b)
    return a+b

vergleiche(1,2)
c = vergleiche(5,4)
print(c)
```

### Klassen

```python
class Person:
    alter = 0
    name = ""

    def sag_deinen_namen(self):
        print("Mein Name ist", self.name)


PersonA = Person()
PersonA.name = "Platon"
PersonA.sag_deinen_namen()
```

#### Vererben

```python
class Schueler(Person):
    turnbeutel = "vergessen"

personB = Schueler()
personB.name = "Aristoteles"
personB.sag_deinen_namen()
print(personB.name, "gesteht", PersonA.name, ": Ich habe den Turnbeutel", 
      personB.turnbeutel)
```
