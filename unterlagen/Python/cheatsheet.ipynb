## Listen

In Listen können mehrere Variablen gespreichert werden. Sie werden mit einer eckigen Klammer deklariert:

```python
# Eine Liste erstellen
meine_liste = ["Apfel", "Banane", "Kirsche"]

# Auf Elemente zugreifen
print(meine_liste[0])   # Erstes Element: Apfel
print(meine_liste[1])   # Zweites Element: Banane
print(meine_liste[-1])  # Letztes Element: Kirsche
print(meine_liste[-2])  # Vorletztes Element: Banane

# Mehrere Elemente auswählen
print(meine_liste[0:2])   # ["Apfel", "Banane"]
print(meine_liste[::-1])  # Liste rückwärts ausgeben
```

## Bedingungen

```python
if x > 0:
    print("Positiv")
elif x == 0:
    print("Null")
else:
    print("Negativ")
```

## Schleifen

```python
for i in meine_liste:
    print(i)
```

```python
# Schleife mit while und einer Zahl als Bedingenung
while x != 0:
    print(x)
    if x > 0:
        x -= 1
    else:
        x += 1
```

## Funktionen

Funktionen fassen Code zusammen:

```python
# Funktionsdefinition
def my_function(a, b):
    if a > b:
        return a
    else:
        return b

# Funktionsaufruf
result = my_function(10, 5)
print("Die größere Zahl ist:", result)
```

### Übergabeparameter
In den runden Klammern steht der Übergabeparamter.

### return
Beendet eine Funktion und gibt einen Wert zurück.


## Klassen
Klassen ermöglichen komplexere Objekte zu definieren.

```python
class Person:
    name = ""
    alter = 0

    def sag_deinen_namen(self):
        print(self.name)
```

### Objekt einer Klasse erstellen

```python
personA = Person()
```

### Attribute
Eigenschaften eines Objekts, z.B. "name". Sie können einfach gesetzt werden und abgefragt werden:

```python
personA.name = "Max"
personA.alter = 30
print("Alter von", personA.name, "ist", personA.alter)
```

### Methoden
Es können aber auch die Klasseninternen Methoden verwendet werden:

```python
personA.sag_deinen_namen()
```

### Vererbung
Eine Klasse kann von einer anderen erben:

```python
class Schueler(Person):
    Turnbeutel = "vergessen"


# Objekt der Klasse Schueler erstellen
SchuelerA = Schueler()
SchuelerA.name = "Platon"
SchuelerA.sag_deinen_namen()
print(SchuelerA.name, "gesteht Sokrates, das er den Turnbeutel", SchuelerA.Turnbeutel, "hat.")
```

### Die __init__ Methode
Mit `__init__` wird ein neues Objekt beim Erstellen initialisiert:

```python
class PersonMitInit(Person):
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter

    def sag_deinen_namen(self):
        print(f"Mein Name ist {self.name}.")


# Bei der Erstellung wird __init__ automatisch aufgerufen
personMitInit = PersonMitInit("Ada", 25)
personMitInit.sag_deinen_namen()
print(personMitInit.alter)
```
